```python
import os, time, json
from octic_ai_agent import OcticAgent
tasks = [
    ('Research Analyst','List three operational risks when deploying an AI agent and explain each briefly.'),
    ('Software Engineer','Design a small retry strategy for a flaky HTTP client.'),
    ('Data Analyst','Explain how you would validate a CSV pipeline for missing values and duplicates.'),
    ('Workflow Operator','Create a five-step release plan for a tested Python agent.'),
]
results=[]
live=os.getenv('OCTIC_KAGGLE_LIVE')=='1'
for role,prompt in tasks:
    if not live:
        results.append({'role':role,'status':'skipped','seconds':0,'output':'Set OCTIC_KAGGLE_LIVE=1 for live evaluation.'})
        continue
    start=time.perf_counter()
    try:
        out=OcticAgent('You are the '+role+'. Be accurate and concise.',name=role.replace(' ','')).start(prompt)
        results.append({'role':role,'status':'passed','seconds':round(time.perf_counter()-start,3),'output':str(out)})
    except Exception as exc:
        results.append({'role':role,'status':'failed','seconds':round(time.perf_counter()-start,3),'output':f'{type(exc).__name__}: {exc}'})
print(json.dumps(results,indent=2))
assert (not live) or all(x['status']=='passed' for x in results), 'One or more live agent evaluations failed.'
```

```python
from pathlib import Path
Path('/kaggle/working/octic_agent_eval_results.json').write_text(json.dumps(results,indent=2))
print('Saved /kaggle/working/octic_agent_eval_results.json')
```
